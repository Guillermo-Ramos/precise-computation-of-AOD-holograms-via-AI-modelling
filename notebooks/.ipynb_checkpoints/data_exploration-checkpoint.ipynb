```python
import scipy.io
import numpy as np

mat = scipy.io.loadmat("../data/example/arxiu_sense_freqs.mat")
```

```python
# 1. Filtrar los metadatos internos y ver solo tus variables
variables = [k for k in mat.keys() if not k.startswith('__')]
print("Variables encontradas:", variables)

# 2. Inspeccionar la forma y tipo de cada variable
for var_name in variables:
    data = mat[var_name]
    if isinstance(data, np.ndarray):
        print(f"\nVariable '{var_name}':")
        print(f"  Tipo: {type(data)} | Dtype: {data.dtype} | Dimensiones: {data.shape}")
        
        # Si no es enorme, ver una muestra de los primeros valores
        if data.size > 0:
            print("  Muestra de datos:")
            print(data)#.flat[:])  # Primeros 5 elementos
    else:
        print(f"\nVariable '{var_name}': {type(data)} -> {data}")
```
